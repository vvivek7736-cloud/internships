```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
iris = sns.load_dataset('iris')
```

```python
iris.head()
```

```python
print("Dataset Shape:", iris.shape)
```

```python
iris.info()
```

```python
iris.describe()
```

```python
print(iris.isnull().sum())
```

```python
print("Duplicate Rows:", iris.duplicated().sum())
```

```python
print(iris['species'].value_counts())
```

```python
sns.countplot(x='species', data=iris)
plt.title("Species Distribution")
plt.show()
```

```python
iris['species'].value_counts().plot.pie(
    autopct='%1.1f%%',
    figsize=(8,8)
)

plt.title("Species Distribution")
plt.ylabel("")
plt.show()
```

```python
iris.hist(figsize=(12,10))
plt.show()
```

```python
plt.figure(figsize=(10,8))

plt.subplot(2,2,1)
plt.hist(iris['sepal_length'])
plt.title('Sepal Length')

plt.subplot(2,2,2)
plt.hist(iris['sepal_width'])
plt.title('Sepal Width')

plt.subplot(2,2,3)
plt.hist(iris['petal_length'])
plt.title('Petal Length')

plt.subplot(2,2,4)
plt.hist(iris['petal_width'])
plt.title('Petal Width')

plt.tight_layout()
plt.show()
```

```python
sns.boxplot(data=iris.drop('species', axis=1))
plt.title("Boxplot of Numerical Features")
plt.show()
```

```python
sns.boxplot(
    x='species',
    y='petal_length',
    data=iris
)

plt.show()
```

```python
sns.stripplot(
    x='species',
    y='petal_length',
    data=iris,
    jitter=True
)

plt.show()
```

```python
sns.scatterplot(
    x='sepal_length',
    y='sepal_width',
    hue='species',
    data=iris
)

plt.show()
```

```python
sns.pairplot(
    iris,
    hue='species'
)

plt.show()
```

```python
iris.drop('species', axis=1).corr()
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    iris.drop('species', axis=1).corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
fig, axes = plt.subplots(2,2, figsize=(12,8))

sns.boxplot(x='species', y='sepal_length', data=iris, ax=axes[0,0])
sns.boxplot(x='species', y='sepal_width', data=iris, ax=axes[0,1])
sns.boxplot(x='species', y='petal_length', data=iris, ax=axes[1,0])
sns.boxplot(x='species', y='petal_width', data=iris, ax=axes[1,1])

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.violinplot(
    x='species',
    y='petal_length',
    data=iris
)

plt.show()
```

```python
from pandas.plotting import andrews_curves

plt.figure(figsize=(10,6))

andrews_curves(
    iris,
    'species'
)

plt.show()
```

```python
from pandas.plotting import parallel_coordinates

plt.figure(figsize=(10,6))

parallel_coordinates(
    iris,
    'species'
)

plt.show()
```
