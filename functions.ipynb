```python
#Create a function to convert a string to uppercase.
def conv(name):
    if(name.strip()==""):
        print("ONLy spaces found!!!")
        return
    if(name.isupper()):
        print("Already uppercased !!!")
        return 
    print(name.upper())
n=input("Enter your string : ")
conv(n)
```

```python
#Create a function to reverse a string.
def conv(name):
    x=""
    for i in name:
        x=i+x
    print("og string is : ",name)
    print("reversed  : ",x)
n=input("Enter your string : ")
conv(n)
```

```python
#Create a function to check whether a string is a palindrome.
def conv(name):
    x=""
    for i in name:
        x=i+x
    print(x)
    if(name==x):
        print("palindrome")
    else:
        print("not palindrome")
n=input("Enter your string : ")
conv(n)
```

```python
#Create a function to count vowels in a string.
def conv(name):
    v="aeiou"
    count=0
    name=name.lower()
    for i in name:
        if i in v:
            count+=1
    print("NUmber of vowelse : ",count)        
n=input("Enter your string : ")
conv(n)
```

```python
#Create a function to separate even and odd numbers into different lists.
a=[10000,2,3,4,5,6,1,2,2,1,1,1,3,984949]
def find(a):
    x=[]
    y=[]
    for i in a:
        if(i%2==0):
            x.append(i)
        else:
            y.append(i)
    return x,y
x,y=find(a)
print(x)
print(y)
```

```python
#Create a function to sort a list without using sort().
a=[10000,2,3,4,5,6,1,2,2,1,1,1,3,984949]
def find(a):
    for i in range(len(a)-1):
        for j in range(i+1,len(a)):
            if(a[i]>a[j]):
                a[i],a[j]=a[j],a[i]
    return a
a=find(a)
print(a)
```

Dictionary-Based Functions

```python
#Create a function to display all keys of a dictionary.
a={
    'name':'vivek',
    'age':20,
    'sex':'male'
}
def keyshow(a):
    print("keys are : ")
    for i in a.keys():
        print(i)
keyshow(a)
```

```python
#Create a function to display all values of a dictionary.
a={
    'name':'vivek',
    'age':20,
    'sex':'male'
}
def keyshow(a):
    print("values are : ")
    for i in a.values():
        print(i)
keyshow(a)
```

```python
#Create a function to add a new key-value pair.
a={
    'name':'vivek',
    'age':20,
    'sex':'male'
}
def show(a):
    x=input("Enter key")
    y=input("Enter value")
    print(f"Newly added pair => {x} : {y}")
    a[x]=y
show(a)
print(a)
```

```python
#Create a function to update a value in a dictionary.
a={
    'name':'vivek',
    'age':20,
    'sex':'male'
}
def show(a):
    x=input("Enter key to be updated ")
    if x in a:
        y=input("Enter new value ")
        if y==str(a[x]):
            print("New value is same as old!!!")
        else:
            a[x]=y
            print(f"Key {x} updated !!!")
    else:
        print("Key NOt found!!")
show(a)
print(a)
```

```python
#Create a function to delete a key from a dictionary.
a={
    'name':'vivek',
    'age':20,
    'sex':'male'
}
def show(a):
    x=input("Enter key to be deleted ")
    rem=a.pop(x,None)
    if rem is None:
        print(f"Key {x} not found!!")
    else:
        print(f"Key {x} rem!!")
show(a)
print(a)
```
