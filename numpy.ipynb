Indexing & Slicing

```python
arr = np.array([1, 2, 3, 4, 5])
arr[0]

arr[1:4]
arr[::2]
```

```python
# 2D Arrays
arr = np.array([[1, 2, 3], [4, 5, 6]])
arr[0, 1]
arr[:, 1]
arr[1, :2]
```

```python
# Boolean Indexing
arr = np.array([1, 2, 3, 4, 5])
mask = arr > 3
arr[mask]
```

Reshaping & Combining Arrays

```python
# Reshape
arr = np.arange(6)
arr.reshape(2, 3)
```

```python
# Concatenation
a = np.array([1, 2])
b = np.array([3, 4])
np.concatenate((a, b))
```

```python
# Stacking
np.vstack((a, b))
np.hstack((a, b))
```

Broadcasting

```python
a = np.array([1, 2, 3])
a + 5 # [6, 7, 8]
```
