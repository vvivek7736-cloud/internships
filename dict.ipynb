Practical Questions on Dictionaries
Basic Level

```python
#Create a dictionary to store a student's name, age, and course.
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
print(stud)
```

```python
#Add a new key-value pair to an existing dictionary.
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
stud['gender']='male'
print(stud)
```

```python
#Update the age of a student in a dictionary.
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
stud['name']='anand'
print(stud)
```

```python
#Delete a specific key from a dictionary.
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
stud.pop("name",None)
print(stud)
```

```python
#Print all keys in a dictionary.
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
print(stud.keys())
```

```python
#Print all values in a dictionary.
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
print(stud.values())
```

```python
#Print all key-value pairs in a dictionary.
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
print(stud)
```

```python
#Check whether a particular key exists in a dictionary.
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
print('course' in stud)
```

```python
#Create a copy of a dictionary.
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
st=stud.copy()
print(stud)
print(st)
```

```python
#Clear all items from a dictionary.
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
stud.clear()
print(stud)
```

Intermediate Level

```python
#Count the frequency of each character in a string.
st=input("Enter string : ")
f={}
for ch in st:
    if ch==' ':
        continue
    f[ch]=f.get(ch,0)+1
print(f)
```

```python
#Count the frequency of words in a sentence.
st=input("Enter string : ")
w=st.split()
f={}
for x in w:
    f[x]=f.get(x,0)+1
print(f)
```
