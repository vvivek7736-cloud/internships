Practical Questions on Python Operators


Basic Level

```python
#Write a program to add two numbers using arithmetic operators.
a=int(input("Enter a :"))
b=int(input("Enter b :"))
s=a+b
print("Sum is ",s)
```

```python
#Write a program to subtract two numbers.
a=int(input("Enter a :"))
b=int(input("Enter b :"))
s=a-b
print("diff is ",s)
```

```python
#Write a program to multiply two numbers.
a=int(input("Enter a :"))
b=int(input("Enter b :"))
s=a*b
print("product is ",s)
```

```python
#Write a program to divide two numbers and display the result.
a=int(input("Enter a :"))
b=int(input("Enter b :"))
if(b==0):
    print("Division not possible")
else:
    ans=a/b
    print("ans is ",ans)
```

```python
#Write a program to find the remainder when one number is divided by another.
a=int(input("Enter a :"))
b=int(input("Enter b :"))
if(b==0):
    print("Division not possible")
else:
    ans=a%b
    print("rem is ",ans)
```

```python
#Write a program to find the square of a number using the exponent operator.
a=int(input("Enter a :"))
b=int(input("Enter b :"))
print(f"{a}^{b} = {a**b}")
```

```python
#Write a program to calculate the area of a rectangle using length and width.
a=int(input("Enter length :"))
b=int(input("Enter bredth :"))
print(f"Area of rec = {a*b}")
```

```python
#Write a program to calculate the perimeter of a rectangle.
a=int(input("Enter length :"))
b=int(input("Enter bredth :"))
print(f"Perimeter of rec = {2*(a+b)}")
```

```python
#Write a program to swap two numbers using a temporary variable.
a=int(input("Enter a :"))
b=int(input("Enter b :"))
print(f"Before swap  : a={a} and b={b}")
temp=a
a=b
b=temp
print(f"After swap  : a={a} and b={b}")
```

```python
#Write a program to find the average of three numbers.
a=int(input("Enter a :"))
b=int(input("Enter b :"))
c=int(input("Enter c :"))
avg=(a+b+c)/3
print("avg is : ",avg)
```

```python
#Write a program to compare the ages of two people and display who is older.
a=int(input("Enter age of a :"))
b=int(input("Enter age of b :"))
if(a>b):
    print("A is elder")
elif(a==b):
    print("Age of A and B equal")
else:
    print("B is elder")
```

```python
#Write a program to determine whether a student has passed (marks ≥ 40).
a=float(input("Enter mark of stud :"))
if(a>=40):
    print("Passed")
else:
    print("Failed")
```

Logical Operator Questions


```python
#Check whether a person's age is between 18 and 60 using and.
a=int(input("Enter age :"))
print(a>18 and a<60)
```

```python
#Check whether a number is divisible by 3 or 5 using or.
a=int(input("Enter number :"))
print(a%3==0 or a%5==0)
```

```python
#Check whether a user is not logged in using not.
a=int(input("Enter number 1-> login:"))
if(a==1):
    a=True
else:
    a=False
print("True if user not login : ",not(a))
```

```python
#Determine whether a student qualifies for a scholarship:Marks above 80Attendance above 75% 
mark,att=map(int,input("Enter mark and att in %:").split())
print("mark :",mark)
print("att :",att,"%")
print(mark>80 and att>75)
```
