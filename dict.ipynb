Practical Questions on Dictionaries
Basic Level

```python
#Create a dictionary to store a student's name, age, and course.
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
print(stud)
```

```python
#Add a new key-value pair to an existing dictionary.
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
stud['gender']='male'
print(stud)
```

```python
#Update the age of a student in a dictionary.
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
stud['name']='anand'
print(stud)
```

```python
#Delete a specific key from a dictionary.
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
stud.pop("name",None)
print(stud)
```

```python
#Print all keys in a dictionary.
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
print(stud.keys())
```

```python
#Print all values in a dictionary.
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
print(stud.values())
```

```python
#Print all key-value pairs in a dictionary.
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
print(stud)
```

```python
#Check whether a particular key exists in a dictionary.
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
print('course' in stud)
```

```python
#Create a copy of a dictionary.
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
st=stud.copy()
print(stud)
print(st)
```

```python
#Clear all items from a dictionary.
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
stud.clear()
print(stud)
```

Intermediate Level

```python
#Count the frequency of each character in a string.
st=input("Enter string : ")
f={}
for ch in st:
    if ch==' ':
        continue
    f[ch]=f.get(ch,0)+1
print(f)
```

```python
#Count the frequency of words in a sentence.
st=input("Enter string : ")
w=st.split()
f={}
for x in w:
    f[x]=f.get(x,0)+1
print(f)
```

```python
#Merge two dictionaries into one.
marks={
    'varun' : 10,
    'arun' : 20,
    'vivek' : 30,
    'anoop' : 15

}
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
stud.update(marks)
print(stud)
```

```python
#Create a dictionary from two lists (one containing keys and one containing values).
stud=[
    'varun' ,
    'arun' ,
    'vivek' ,
    'anoop' 
]
mark=[10,20,30,40]
d={}
d=dict(zip(stud,mark))
print(d)
```

```python
#Store employee details using nested dictionaries.
emp={
    'emp1' : {'name' : 'vivek' , 'branch' : 'main' , 'pos' : 'ceo'},
    'emp2' : {'name' : 'arun' , 'branch' : 'hr' , 'pos' : 'manager'},
    'emp3' : {'name' : 'varun' , 'branch' : 'cr' , 'pos' : 'assistent'}
}
print(emp)
```

```python
#Find all keys whose values are greater than 50.
s={1:10,2:200,3:30,4:80,5:50,6:60,7:89}
print(s)
for x in s:
    if s[x] > 50:
        print(x)
```

```python
#Create a dictionary where keys are numbers from 1 to 10 and values are their squares.
s={}
for i in range(1,11):
    s[i]=i*i
print(s)
```

```python
#Sort a dictionary by its values.
p={1:10,2:200,3:30,4:80,5:50,6:60,7:89}
s=sorted(p.items(),key=lambda x : x[1])
print(dict(s))
```

```python
#Convert a list of names into a dictionary where the value is the length of each name.
name=['vivek','arun','varun','anoop']
d={}
for x in name:
    d[x]=len(x)
print(d)
```
