```python
#Create a dictionary to store a student's name, age, and course.
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
print(stud)
```

```python
#Delete a specific key from a dictionary.
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
stud.pop("name",None)
print(stud)
```

```python
#Print all keys in a dictionary.
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
print(stud.keys())
```

```python
#Print all values in a dictionary.
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
print(stud.values())
```

```python
#Print all key-value pairs in a dictionary.
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
print(stud)
```

```python
#Check whether a particular key exists in a dictionary.
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
print('course' in stud)
```

```python
#Create a copy of a dictionary.
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
st=stud.copy()
print(stud)
print(st)
```

```python
#Clear all items from a dictionary.
stud={
    'name' : 'vivek',
    'age' : 20,
    'course' : 'aiml'
}
stud.clear()
print(stud)
```

```python
#Count the frequency of each character in a string.
st=input("Enter string : ")
f={}
for ch in st:
    if ch==' ':
        continue
    f[ch]=f.get(ch,0)+1
print(f)
```

```python
#Count the frequency of words in a sentence.
st=input("Enter string : ")
w=st.split()
f={}
for x in w:
    f[x]=f.get(x,0)+1
print(f)
```

```python
#Sort a dictionary by its values.
p={1:10,2:200,3:30,4:80,5:50,6:60,7:89}
s=sorted(p.items(),key=lambda x : x[1])
print(dict(s))
```

```python
#Convert a list of names into a dictionary where the value is the length of each name.
name=['vivek','arun','varun','anoop']
d={}
for x in name:
    d[x]=len(x)
print(d)
```

```python
#Build a contact book using a dictionary.
con={}
num=set()
def add():
    x=input("Enter name : ")
    if x in con:
        print("This name already exist!!")
    else:
        y=input("Enter number: ")
        if y in num:
            print("This number already exist!!")
        else:
            con[x]=y
            num.add(y)
            print(f"{x} succesfully added !!")
def delete():
    x=input("Enter name : ")
    rem=con.pop(x,None)
    if rem is None:
        print("Name not exist!!")
    else:
        print(f"Name {x} deleted!!")
        num.discard(rem)
def view():
    if not con:
        print("contact book is empty !!!")
    else:
        print("All contacts : ")
        for i in con:
            print(f"{i} : {con[i]}")
def search():
    x=input("Enter name : ")
    if x in con:
        print(f"{x} found!!")
        print(f"{x} : {con[x]}")
    else:
        print(f"{x} not found!!")
while(1):
    print("====Contact Boook===")
    print("1.add\n2.delete\n3.view all\n4.search\n5.exit\n")
    c=int(input("Enter your choice : "))
    if(c==1):
        add()
    elif(c==2):
        delete()
    elif(c==3):
        view()
    elif(c==4):
        search()
    elif(c==5):
        print("Exiting.....")
        break
    else:
        print("Invalid choice!!!!")
```

```python
#Store product names and prices in a dictionary and calculate the total bill.
bill={}
def enter():
    a=input("Enter product name: ")
    b=int(input("Enter product price: "))
    q=int(input("Enter product qty: "))
    bill[a]={'price':b,'qty':q}
def display():
    for x,y in bill.items():
        print(f"{x} : {y['price']} x {y['qty']}")
def cal():
    print("calclating the toal price....")
    tot=0
    for i in bill.values():
        tot+=i['price']*i['qty']
    print("Total price : ",tot)
while(1):
    c=int(input("Enter : "))
    if(c==1):
        enter()
    elif(c==2):
        display()
    elif(c==3):
        cal()
        break
    else:
        print("invalid!!!")
```

```python
#Count the occurrence of each word in a paragraph.
para={}
p=input("Enter para : ")
words=p.split()
for w in words:
    para[w]=para.get(w,0)+1
for i in para:
    print(f"{i} : {para[i]}")
```

```python
#Create a student result system that stores marks for multiple subjects and calculates the total and average marks
stud={}
n=int(input("Enter no of stud: "))
for i in range(n):
    name=input("Enter stud name: ")
    markx=int(input("Enter mark1: "))
    marky=int(input("Enter mark2: "))
    markz=int(input("Enter mark3: "))
    totx=markx+marky+markz
    avgx=totx/3
    stud[name]={'mark1':markx,'mark2':marky,'mark3':markz,'tot':totx,'avg':avgx}
for i in stud:
    print(f"{i} : {stud[i]}")
```
