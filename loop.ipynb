Basic Level

```python
#Print numbers from 1 to 10 using a for loop.
for i in range(1,11):
    print(i)
```

```python
#Print numbers from 10 to 1 using a for loop.
for i in range(10,0,-1):
    print(i)
```

```python
#Print all even numbers between 1 and 50.
for i in range(2,51,2):
    print(i)
```

```python
#Print all odd numbers between 1 and 50.
for i in range(1,51,2):
    print(i)
```

```python
#Print the squares of numbers from 1 to 10.
for i in range(1,11):
    print(f" {i}^2 = ",i*i)
```

```python
#Print the cubes of numbers from 1 to 10.
for i in range(1,11):
    print(f" {i}^{3} = ",i*i*i)
```

```python
#Find the sum of numbers from 1 to 100.
sum=0
for i in range(1,101):
    sum+=i
print(f" sum = ",sum)
```

```python
#Find the sum of all even numbers between 1 and 100.
tot=0
for i in range(2,101,2):
    tot+=i
print(f" sum of even no:s = ",tot)
```

```python
#Find the sum of all odd numbers between 1 and 100.
tot=0
for i in range(1,101,2):
    tot+=i
print(f" sum of odd no:s = ",tot)
```

```python
#Print the multiplication table of a given number.
tot=0
num=int(input("Enter a number to show multiplication table : "))
for i in range(1,11):
    print(f" {num} x {i} = {num*i}")
```
