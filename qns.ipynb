```python
#Create a grade calculator based on marks.
marks = []

# input marks for 5 subjects
for i in range(5):
    m = int(input(f"Enter mark {i+1}: "))
    marks.append(m)

total = sum(marks)
percentage = (total / 500) * 100

print("Total:", total)
print("Percentage:", percentage)

# grading system
if percentage >= 90:
    print("Grade: A+")
elif percentage >= 80:
    print("Grade: A")
elif percentage >= 70:
    print("Grade: B")
elif percentage >= 60:
    print("Grade: C")
elif percentage >= 50:
    print("Grade: D")
else:
    print("Grade: F")
```
